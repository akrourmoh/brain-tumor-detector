# 01 — Data Exploration
**Objectif :** Comprendre le dataset avant de toucher au code.
- Compter les images par classe
- Visualiser des exemples
- Vérifier les tailles d'images
- Constater le déséquilibre des classes

```python
import os
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

## 1. Compter les images par classe

```python
data_dir = 'data'

yes_count = len(os.listdir(os.path.join(data_dir, 'yes')))
no_count  = len(os.listdir(os.path.join(data_dir, 'no')))

print(f'Images avec tumeur (yes) : {yes_count}')
print(f'Images sans tumeur (no)  : {no_count}')
print(f'Total                    : {yes_count + no_count}')
print(f'Ratio yes/no             : {yes_count/no_count:.2f}')
```

## 2. Visualiser des exemples

```python
def show_samples(folder, label, n=5):
    files = os.listdir(folder)[:n]
    fig, axes = plt.subplots(1, n, figsize=(15, 3))
    fig.suptitle(f'Classe : {label}', fontsize=14)
    for i, f in enumerate(files):
        img = cv2.imread(os.path.join(folder, f))
        img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
        axes[i].imshow(img)
        axes[i].axis('off')
    plt.tight_layout()
    plt.show()

show_samples(os.path.join(data_dir, 'yes'), 'Tumeur (yes)')
show_samples(os.path.join(data_dir, 'no'),  'Pas de tumeur (no)')
```

## 3. Vérifier les tailles d'images

```python
shapes = []
for label in ['yes', 'no']:
    folder = os.path.join(data_dir, label)
    for f in os.listdir(folder):
        img = cv2.imread(os.path.join(folder, f))
        if img is not None:
            shapes.append(img.shape)

unique_shapes = set(shapes)
print(f'Nombre de tailles différentes : {len(unique_shapes)}')
print(f'Exemples de tailles : {list(unique_shapes)[:10]}')
print()
print('=> Les images ont des tailles différentes : le resize est indispensable.')
```

## 4. Visualiser le déséquilibre des classes

```python
fig, ax = plt.subplots(figsize=(6, 4))
ax.bar(['yes (tumeur)', 'no (sain)'], [yes_count, no_count], color=['tomato', 'steelblue'])
ax.set_title('Distribution des classes — dataset original')
ax.set_ylabel('Nombre d\'images')
for i, v in enumerate([yes_count, no_count]):
    ax.text(i, v + 1, str(v), ha='center', fontweight='bold')
plt.tight_layout()
plt.show()

print('=> Déséquilibre : 155 vs 98. La data augmentation va corriger ça.')
```
